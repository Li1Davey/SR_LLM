```python

def plot_one_metric(total, name, title, metric_name, show_title=False, show_ylabel=True,use_abs=True):
    GP=[float(line.split("\t")[0]) for line in total.split("\n")]
    CVGP=[float(line.split("\t")[1]) for line in total.split("\n")]
    VPG=[float(line.split("\t")[2]) for line in total.split("\n")]
    PQT=[float(line.split("\t")[3]) for line in total.split("\n")]
    DSR=[float(line.split("\t")[4]) for line in total.split("\n")]
    GPMELD=[float(line.split("\t")[5]) for line in total.split("\n")]
    Eureqa=[float(line.split("\t")[6]) for line in total.split("\n")]
    MCTS=[float(line.split("\t")[7]) for line in total.split("\n")]
    CVMCTS=[float(line.split("\t")[8]) for line in total.split("\n")]
    full_data=CVGP+GP+Eureqa+CVMCTS+MCTS+VPG+PQT+DSR+GPMELD
    if use_abs:
        full_data=[abs(x) for x in full_data]
    df=pd.DataFrame({
        "model": full_data,
        "type": ['CvGP(ours)']*10+['GP']*10+['Eureqa']*10+['CvMCTS(ours)']*10+['MCTS']*10+['VPG']*10+['PQT']*10+['DSR']*10+['GPMeld']*10
    })
    

    plt.figure(figsize=(3,3))
    ax=sns.boxplot(data=df, x="model", y="type", 
                   showfliers=True, 
                   width=0.5, hue="type",
                    palette=palette)
    
    
    if show_title==True:
        ax.set_title(title, fontsize=16)
    
    ax.set(xlabel=metric_name, ylabel='')
    if logx==True:
        plt.xscale('log', base=10)
    if show_ylabel==False:
        ax.set(yticklabels=[])
#     plt.xticks(rotation=90)
    plt.yticks(rotation=40)
    # ax.xaxis.set_major_locator(plt.MaxNLocator(10))
    fname = os.path.join(basepath, name+"_metric.pdf")
    plt.savefig(fname, bbox_inches='tight', pad_inches=0)
```
