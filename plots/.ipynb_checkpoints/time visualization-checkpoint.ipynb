```python

import seaborn as sns
import matplotlib
%matplotlib inline
import matplotlib.pyplot as plt
from matplotlib import rc
palette=['#ff3b30', '#007aff', '#4cd964','#ffcc00', '#5856d6']
markers=['o', '^', '<', 'd', '*','2']
sns.set_style("ticks")
import matplotlib as mpl
mpl.rcParams['lines.markersize'] = 6
mpl.rcParams['lines.linewidth']=2
rc("font", **{'family':'serif','serif':['Palatino Linotype'], 'size':12})
import os
import pandas as pd
```

```python
basepath="/home/jiangnan/PycharmProjects/scibench/plots"
```

```python

def plot_gp_egp(gp_times,egp_time, name, title, use_logy=False):
    df=pd.DataFrame({
        "model": egp_time+gp_times,
        "type": ['EGP']*len(egp_time)+['GP']*len(gp_times)
    })

    plt.figure(figsize=(4,3))
    ax=sns.boxplot(data=df, y="model", x="type", showfliers=False,
                    palette=palette[:5])
    ax.set_title(title, fontsize=15)
    ax.set(ylabel="Empirical Running Time (sec)", xlabel="Methods")
    if use_logy==True:
        plt.yscale('log', base=10)
    fname = os.path.join(basepath, name+".pdf")
    plt.savefig(fname, bbox_inches='tight', pad_inches=0)
```
