```python

def plot_gp_egp(gp_times,egp_time, name, title, use_logy=False):
    df=pd.DataFrame({
        "model": egp_time+gp_times,
        "type": ['CVGP']*len(egp_time)+['GP']*len(gp_times),
        "generation":[i for i in range(len(egp_time))]+[i for i in  range(len(gp_times))]
    })

    plt.figure(figsize=(4,3))
    ax=sns.lineplot(data=df, y="model", x="generation", hue="type", markers=True,
                    palette=palette[:5])
    ax.set_title(title, fontsize=15)
    ax.set(ylabel="Empirical Running Time (sec)", xlabel="Generations")
    if use_logy==True:
        plt.yscale('log', base=10)
    fname = os.path.join(basepath, name+".pdf")
    plt.savefig(fname, bbox_inches='tight', pad_inches=0)
```
