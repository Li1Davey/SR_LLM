```python
import os
import json
```

```python
selected=[]
for root, dirs, files in os.walk(".", topdown=False):
    for name in files:
        if "Livermore2_Vars4" in name:
            selected.append(os.path.join(root, name))
```

```python
for fi in range(0,10):
    fr=open(f"../equations_trigometric/sincos_nv5_nt55_prog_"+str(fi)+".in", "r").read()
    content=fr.split("\n")[-1]
    j_cont=json.loads(content)
    eq=j_cont['expr'].lower().replace("**", "^").replace("*", "").replace("sin", "\sin").replace("cos", "\cos").replace("exp", "\exp").replace("log", "\log").replace("sqrt","\sqrt")
    
    print(r" {} & $5$ & ${}$ \\ \midrule".format(fi+1, eq))
```
