```python

def plot_gp_egp(total, name, title):
    GP=[abs(float(line.split("\t")[0])) for line in total.split("\n")]
    EGP=[abs(float(line.split("\t")[1])) for line in total.split("\n")]
    df=pd.DataFrame({
        "model": EGP+GP,
        "type": ['EGP']*10+['GP']*10
    })

    plt.figure(figsize=(4,3))
    ax=sns.boxplot(data=df, y="model", x="type", showfliers=False,
                    palette=palette[:5])
    ax.set_title(title, fontsize=15)
    ax.set(ylabel="Normalized MSE Metric", xlabel="Methods")
    fname = os.path.join(basepath, name+".pdf")
    plt.savefig(fname, bbox_inches='tight', pad_inches=0)
```
